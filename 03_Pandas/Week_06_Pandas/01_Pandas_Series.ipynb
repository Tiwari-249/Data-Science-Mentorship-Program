A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

#### **Copy and View**

**1. Copy**

The `copy()` method creates a separate copy of a Pandas Series or DataFrame. Changes made to the copied object generally do not affect the original object.

Example - `df2 = df.copy()`

- df2 is a separate copy of df.

- Changes in df2 do not affect df.

Types of copy

- Shallow copy: `df.copy(deep=False)` — shares underlying data in ways that can allow changes to be reflected across objects.

- Deep copy: `df.copy(deep=True)` — copies the data and index structure; this is the default.

**2. View**

A `view()` is an object that shares data with the original object rather than creating a fully independent copy.

- Changes to shared data may affect both objects.

- Pandas does not provide a general-purpose DataFrame.view() method like NumPy does.

- In Pandas, views can occur during operations such as selecting data, depending on the operation and memory layout.



```python
# Assign top head values of vk series to a new reference variable (shares original memory)
new = vk.head()
```

```python
# Update value in the shared memory reference
new[1] = 1
```

```python
# Display the modified view of the dataset
new
```

```python
# Display original series, confirming it also changed due to modifying the view
vk
```

```python
# Create an independent clone of the data slice using copy()
new = vk.head().copy()
```

```python
# Update elements in the independent copied series
new[1] = 100
```

```python
# Display copied series with updated records
new
```

```python
# Confirm original series remains unaffected after modifying a copied object
vk
```
