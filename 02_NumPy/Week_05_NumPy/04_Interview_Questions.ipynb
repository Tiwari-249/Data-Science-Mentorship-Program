# **meshgrids**

Meshgrid converts 1D coordinate arrays into 2D coordinate matrices so that mathematical functions can be evaluated on a grid of points.

It is a function in NumPy used to create coordinate matrices from coordinate vectors.

It is commonly used for evaluating functions on a grid, plotting 3D surfaces, and contour plots.

**Syntax** - **`X, Y = np.meshgrid(x, y)`**

- x → values for x-axis

- y → values for y-axis

Returns two 2D arrays:

- X → repeated rows of x values

- Y → repeated columns of y values


````python
# Import the NumPy library as 'np' for numerical computing and array manipulations
import numpy as np
````

````python
# Generate 1D coordinate vectors 'a' and 'b' ranging from -10 to 9 containing 20 evenly spaced points
a = np.linspace(-10,9,20)
b = np.linspace(-10,9,20)
````

````python
# Display the elements of the 1D array 'a'
a
````

````python
# Display the elements of the 1D array 'b'
b
````

````python
# Create a 2D coordinate grid (xx, yy) from the 1D coordinate vectors 'a' and 'b'
xx,yy = np.meshgrid(a,b)
````

````python
# Display the coordinate matrix 'xx' representing the X-coordinates on the grid
xx
````

````python
# Display the coordinate matrix 'yy' representing the Y-coordinates on the grid
yy
````

````python
# Import matplotlib and create a scatter plot of the grid points (xx, yy) to visualize the coordinate grid
import matplotlib.pyplot as plt
plt.scatter(xx,yy)
````

````python
# Define a 3D paraboloid function f(x,y) = x^2 + y^2 and calculate the Z values (zz) for our meshgrid points
def para(x,y):
  return x**2 + y**2

zz = para(xx,yy)
zz
````

````python
# Import Plotly libraries to create and visualize an interactive 3D surface plot of the calculated zz values
import plotly.express as px
import plotly.graph_objects as go

fig = px.scatter_3d()
fig.add_trace(go.Surface(x=xx,y=yy,z=zz))
fig.show()
````

````python
# Define a function containing a natural logarithm and calculate zz. This will trigger warnings/NaNs for negative x values
def log(x,y):
  return 3*np.log(x) + 2*y

zz = log(xx,yy)
zz
````

````python
# Attempt to plot the new 3D surface featuring NaNs and infinite values using Plotly
import plotly.express as px
import plotly.graph_objects as go

fig = px.scatter_3d()
fig.add_trace(go.Surface(x=xx,y=yy,z=zz))
fig.show()
````

# **Working with random**

NumPy provides a random module (numpy.random) used to generate random numbers, shuffle data, and perform random sampling.

````python
# Generate a 3D array of random floating-point numbers of shape (2, 3, 2) in the range [0.0, 1.0)
np.random.random((2,3,2))
````

## **1. randint()**

**Purpose**: Generate random integers within a given range.

- Generates random integers.

- Upper limit is not included.

**Syntax**: `np.random.randint(low, high, size)`

- low → starting value (inclusive)

- high → ending value (exclusive)

- size → shape of output array

Random integers are generated between low (inclusive) and high (exclusive).

````python
# Generate 12 random integers between 1 and 100 and reshape them into a 3x4 2D matrix
np.random.randint(1,100,12).reshape(3,4)
````

## **2. seed()**

**Purpose**: Set the starting point of the random number generator.

If you set the same seed, you get the same random numbers every time.

**Why used**

- Reproducibility in machine learning experiments

- Debugging

- Testing models

Setting a seed fixes the internal state so the same sequence of random numbers can be generated repeatedly.

````python
# Seed the random number generator with 0 to ensure the sequence of generated integers is reproducible
np.random.seed(0)
np.random.randint(1,100,12).reshape(3,4)
````

````python
# Re-apply the same seed of 0 to demonstrate that we get the exact same random numbers as the previous run
np.random.seed(0)
np.random.randint(1,100,12).reshape(3,4)
````

````python
# Re-apply seed of 0 again to verify reproducibility a third time
np.random.seed(0)
np.random.randint(1,100,12).reshape(3,4)
````

## **3. shuffle()**

**Purpose**: Randomly rearrange elements of an array.

**Syntax**: `np.random.shuffle(arr)`

It modifies the original array (in-place) and does not return a new array.

````python
# Create a 1D NumPy array with specific numeric elements and print it
a = np.array([12,41,33,67,89,100])
print(a)
````

````python
# Shuffles the elements of array 'a' in-place and prints the modified, randomized array
np.random.shuffle(a)
print(a)
````

## **4. choice()**

**Purpose**: Randomly select elements from an array.

**Syntax**: `np.random.choice(a, size=None, replace=True, p=None)`

- a → array or integer

- size → number of samples

- replace → sampling with/without replacement

- p → probability distribution

The function randomly selects elements from a given array or range.

````python
# Randomly select 3 elements from array 'a' without replacement (no element can be selected twice)
np.random.choice(a,3,replace=False)
````

# **Structured Arrays**

A structured array in NumPy is an array where each element contains multiple named fields with different data types, similar to a row in a table or a C structure.

````python
# Create an unstructured array containing mixed data types (coerced into a single string dtype)
a = np.array([1,'hello',True,1.5])
a
````

````python
# Try to divide the first element (string '1') by 100, which results in a TypeError
a[0] / 100
````

````python
# Define a structured data type with custom field names and distinct data types (string, integer, float)
dt = np.dtype([
    ('name','U20'),
    ('iq',np.int32),
    ('cgpa',np.float64),
    ('placed','U20')
])
````

````python
# View the structured dtype schema definition
dt
````

````python
# Initialize a structured array representing student records using the defined structured dtype
stu = np.array([
    ('nitish',100,6.66,'Yes'),
    ('ankit',120,8.9,'Yes'),
    ('rahul',80,7.3,'No')
],dtype=dt)

stu
````

````python
# Access a single column field ('placed') from the structured array
stu['placed']
````

````python
# Save the structured array as a reusable binary file named 'student.npy'
np.save('student.npy',stu)
````

# **`np.swapaxes(arr, axis1, axis2)`**
```
Interchange two axes of an array.

Syntax : numpy.swapaxes(arr, axis1, axis2)

arr : [array_like] input array.
axis1 : [int] First axis.
axis2 : [int] Second axis.
Return : [ndarray]
```

````python
# Interchange the axes (transposing the 2D array) using np.swapaxes to swap rows and columns
x = np.array([[1,2,3],[4,5,6]])
print(x)
print(x.shape)
print("Swapped")
x_swapped = np.swapaxes(x,0,1)
print(x_swapped)
print(x_swapped.shape)
````

Note: It is not same as reshaping.

````python
# Change the dimensions of the array 'x' using reshape to show how it differs from swapping axes
x_reshaped = x.reshape(3,2)
print(x_reshaped)
````

# **`np.count_nonzero(arr, axis=None)`**

This function counts the number of non-zero values in the array
https://numpy.org/doc/stable/reference/generated/numpy.count_nonzero.html

```
Syntax : numpy.count_nonzero(arr, axis=None)

Parameters :
    arr : [array_like] The array for which to count non-zeros.

    axis : [int or tuple, optional] Axis or tuple of axes along which to count non-zeros. Default is None, meaning that non-zeros will be counted along a flattened version of arr.

    keepdims : [bool, optional] If this is set to True, the axes that are counted are left in the result as dimensions with size one.

Return : [int or array of int] Number of non-zero values in the array along a given axis. Otherwise, the total number of non-zero values in the array is returned.
```


````python
# Count the number of non-zero elements in a 2D array globally, along axis 0, and along axis 1
a = np.array([[0, 1, 7, 0],
              [3, 0, 2, 19]])

print(np.count_nonzero(a))
print(np.count_nonzero(a, axis=0))
print(np.count_nonzero(a, axis=1))
````

# **`np.tile(A, reps)`**

Construct an array by repeating `A` the number of times given by `reps`.
If reps has length `d`, the result will have dimension of `max(d, A.ndim)`.
```
Parameters:

    A: array_like
        The input array.

    reps: array_like
        The number of repetitions of A along each axis.

Returns
    c: ndarray
        The tiled output array.
```
https://numpy.org/doc/stable/reference/generated/numpy.tile.html

````python
# Use np.tile to repeat the entire 1D array twice sequentially
a = np.array([0, 1, 2])
print(a)
print("Tiled")
print(np.tile(a, 2))
````

````python
# Tile the 1D array 'a' into a 2x2 repetition pattern (two repeats along rows and columns)
np.tile(a, (2, 2))
````

````python
# Tile the 1D array 'a' with 2 repetitions vertically (axis-0) and 3 repetitions horizontally (axis-1)
np.tile(a, (2, 3))
````

# **`np.repeat(a, repeats, axis=None)`**

Repeat elements of an array. `repeats` parameter says no of time to repeat

```
Parameters:
            a: array_like
                    Input array.

            repeats: int or array of ints
                    The number of repetitions for each element. repeats is broadcasted to fit the shape of the given axis.

            axis: int, optional
                    The axis along which to repeat values. By default, use the flattened input array, and return a flat output array.

Returns:
    repeated_array: ndarray
                    Output array which has the same shape as a, except along the given axis.
```
https://numpy.org/doc/stable/reference/generated/numpy.repeat.html

````python
# Use np.repeat to duplicate each individual element of array 'x' two times consecutively
x = np.array([[1,2],[3,4]])
print(x)
print(np.repeat(x, 2))
````

````python
# Repeat individual elements 3 times horizontally along columns (axis=1)
print(x)
print(np.repeat(x, 3, axis=1))
````

````python
# Repeat individual elements 3 times vertically along rows (axis=0)
print(np.repeat(x, 3, axis=0))
````

# **`np.allclose`**

Returns True if two arrays are element-wise equal within a tolerance.

The tolerance values are positive, typically very small numbers. The relative difference `(rtol * abs(b))` and the absolute difference `atol` are added together to compare against the `absolute difference` between `a and b`.

If the following equation is element-wise True, then `allclose` returns `True`.

> `absolute(a - b) <= (atol + rtol * absolute(b))`

```
Syntax : numpy.allclose(arr1, arr2, rtol, atol, equal_nan=False)

Parameters :
    arr1 : [array_like] Input 1st array.
    arr2 : [array_like] Input 2nd array.
    rtol : [float] The relative tolerance parameter.
    atol : [float] The absolute tolerance parameter.
    equal_nan : [bool] Whether to compare NaN’s as equal.
            If True, NaN’s in arr1 will be considered equal to NaN’s in arr2 in the output array.

Return : [ bool] Returns True if the two arrays are equal within the given tolerance, otherwise it returns False.
```
https://numpy.org/doc/stable/reference/generated/numpy.allclose.html
https://www.geeksforgeeks.org/numpy-allclose-in-python/


````python
# Compare two arrays element-wise within floating point tolerances using np.allclose
a = np.array([1.1, 1.2, 1.0001])
print(a)

b = np.array([1., 1.02, 1.001])
print(b)

print(np.abs(a-b))

print(np.allclose(a,b))
print(np.allclose(a,b, atol=0.2))
````

````python
# Compare arrays containing NaN values with np.allclose with and without equal_nan parameter
print(np.allclose([1.0, np.nan], [1.0, np.nan]))
print(np.allclose([1.0, np.nan], [1.0, np.nan], equal_nan=True))
````
