```python
# Import the time module to measure execution time
import time

# Record the start time before the loop begins
start = time.time()

# Loop to print numbers from 1 to 4
for i in range (1,5):
  print (i)

# Record the end time after the loop finishes
end = time.time()

# Calculate and print the elapsed time (end_time - start_time)
print (end - start)
```

```python
# Any Integer Input  to String

# Get integer input from the user
number = int (input('Enter any number : '))

# Define string of digits for conversion
digits = '0123456789'
result = ''

# Convert number to string digit by digit using modulo and floor division
while number !=0:
  # Get the last digit (number % 10) and prepend it to the result string
  result = digits [ number % 10 ] + result
  # Remove the last digit from the number
  number = number // 10

print(result)


# Time Complexity = Log n (as the number of digits is proportional to log base 10 of the number)
```

```python
L = [1,2,3,4]

sum = 0
# Calculate the sum of elements in the list
for i in L:
  sum = sum + i

product = 1

# Calculate the product of elements in the list
for i in L:
  product = product * i

print (sum,product)


# Time Complexity = Order of (n) (due to two separate loops, each iterating n times)
```

```python
A = [1,2,3,4]
B = [5,6,7,8]

# Outer loop iterates through elements of list A
for i in A:
    # Inner loop iterates through elements of list B for each element in A
    for j in B:
      print (i,j)


# Tme Complexity = Order of (n*m) or Order of (n^2) if n and m are of similar size
```

```python
# Reverse list

L = [1,2,3,4,5]

# Iterate only up to the middle of the list
for i in range (0,len(L)//2):
  # Calculate the index of the element at the opposite end
  other = len(L) - i - 1
  # Swap elements: temp = L[i], L[i] = L[other], L[other] = temp
  temp = L[i]
  L[i] = L[other]
  L[other] = temp

print (L)


# Time Complexity = Order of (n) (as it iterates through half the list elements)
```

```python
# This code block demonstrates a calculation related to division, not a standard modulo operation.
# Modulo operation (simulated if b is positive)

a = 10
b = 3

# Handle division by zero or non-positive b to prevent errors
if b<= 0:
  print(-1)

# Perform integer division (floor division)
div = a//b

# Calculate a result based on division and subtraction
# Note: This is not the standard mathematical definition of a modulo operation for all cases.
print (a-div-b)


# Time Complexity = Order of (1) - Constant time operations
```

```python
# Sum of all digit

n =  1903
sum = 0

# Calculate the sum of digits using modulo and floor division
while n > 0:
  # Add the last digit to sum
  sum = sum + n % 10
  # Remove the last digit from the number
  n = n // 10

print (sum)


# Time Complexity = Log n (as the number of iterations is proportional to the number of digits in n)
```

```python
# Fibonacci Series using recursion

def fib(n):
  # Base cases for Fibonacci sequence
  if n == 1 or n == 0:
    return 1
  else:
    # Recursive call to sum the previous two Fibonacci numbers
    return fib(n-1) + fib(n-2)


# Time Complexity = 2^n (due to overlapping subproblems and redundant calculations in naive recursive implementation)
```
