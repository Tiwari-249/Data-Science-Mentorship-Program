## **1. `np.sort`**

Return a sorted copy of an array.

https://numpy.org/doc/stable/reference/generated/numpy.sort.html

```python
# Create a 1D array 'a' with 15 random integers between 1 and 99
a = np.random.randint(1,100,15)
print(a)
```

```python
# Create a 2D array 'b' (6x4) with 24 random integers between 1 and 99
b = np.random.randint(1,100,24).reshape(6,4)
print(b)
```

```python
# Sort array 'a' in ascending order
np.sort(a)
```

```python
# Sort array 'a' in descending order
np.sort(a)[::-1]
```

```python
# Sort array 'b' along the last axis (rows) in ascending order
np.sort(b)
```

```python
# Sort array 'b' along the last axis (rows) and then reverse the order of rows
np.sort(b)[::-1]
```

```python
# Sort array 'b' along axis 0 (columns) in ascending order
np.sort(b,axis=0)
```

```python
# Sort array 'b' along axis 0 (columns) and then reverse the order of rows
np.sort(b,axis=0)[::-1]
```

```python
# Sort array 'b' along axis 1 (rows) in ascending order
np.sort(b,axis=1)
```

```python
# Sort array 'b' along axis 1 (rows) and then reverse the order of columns
np.sort(b,axis=1)[::-1]
```

## **3. `np.concatenate`**

numpy.concatenate() function concatenate a sequence of arrays along an existing axis.

https://numpy.org/doc/stable/reference/generated/numpy.concatenate.html

```python
# Create a 2D array 'c' (2x3) with values from 0 to 5
c = np.arange(6).reshape(2,3)
print(c)
```

```python
# Create a 2D array 'd' (2x3) with values from 6 to 11
d = np.arange(6,12).reshape(2,3)
print(d)
```

```python
# Concatenate arrays 'c' and 'd' along axis 0 (vertically)
np.concatenate((c,d),axis=0)
```

```python
# Concatenate arrays 'c' and 'd' along axis 1 (horizontally)
np.concatenate((c,d),axis=1)
```

## **6. `np.where`**

The numpy.where() function returns the indices of elements in an input array where the given condition is satisfied.

https://numpy.org/doc/stable/reference/generated/numpy.where.html

```python
# Print array 'a'
print(a)
```

```python
# Find all indices with value greater than 50
np.where(a > 50)
```

```python
# Replace all values > 50 with 0
np.where(a > 50, 0, a)
```

```python
# Replace all even number with 0
np.where(a % 2 == 0, 0, a)
```

### **13. `np.corrcoef`**
