# **Tuples**

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

**Characterstics**

- Ordered

- Immutable (Unchangeble)

- Allows duplicate

## Plan of attack

- Creating a Tuple
- Accessing items
- Editing items
- Adding items
- Deleting items
- Operations on Tuples
- Tuple Functions

## **Creating Tuples**

```python
# Empty
t1 = ()

print(t1)
print(type(t1))
```

```python
# Create a tuple with a single item
t2 = (2)

print(t2)
print(type(t2))
```

```python
# This demonstrates that a single item in parentheses without a trailing comma is treated as its original type (string in this case), not a tuple
t2 = ('hello')

print(t2)
print(type(t2))
```

```python
# Comma is mandatory for creating a tuple with a single item
t2 = (2,)

print(t2)
print(type(t2))
```

```python
# A trailing comma is required to create a tuple with a single item
t2 = ('hello',)

print(t2)
print(type(t2))
```

```python
# Homogenous
t3 = (1,2,3,4)
print(t3)
```

```python
# Hetrogenous
t4 = (1,2.5,True,[1,2,3])
print(t4)
```

```python
# 2D Tuple
t5 = (1,2,(3,4,5))
print(t5)
```

```python
# Using type conversion
t6 = tuple('hello')
print(t6)
```

## Difference between Lists and Tuples

### **Speed**

```python
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*5
print('List time: ',time.time()-start)

start = time.time()
for i in T:
  i*5
print('Tuple time: ',time.time()-start)
```

### **Memory**

```python
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size: ',sys.getsizeof(L))
print('Tuple size: ',sys.getsizeof(T))
```

### **Error prone**

```python
# List
a = [1,2,3]
b = a

a.append(4)

print(a)
print(b)
```

```python
# Tuple
a = (1,2,3)

b = a

a = a + (4,)

print(a)
print(b)
```

# **Sets**

A set is an **unordered collection** of items. Every set element is **unique** (no duplicates) and must be **immutable** (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

**Characterstics**

- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

## **Creating Sets**

```python
# Empty Set
s = {}

print(s)
print(type(s))
```

```python
s = set({})
print(s)
print(type(s))
```

```python
# 2D Set
s2 = {1,2,3,{4,5}}
print(s3)
```

# **Dictionary**

Dictionary in Python is a collection of **keys values**, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as **map** or **assosiative arrays**.

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }

**Characterstics**

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

## **Create Dictionary**

```python
# Empty dictionary
d = {}

print(d)
print(type(d))
```

```python
# 1D dictionary --> Homogenous
d1 = { 'name' : 'nitish' ,'gender' : 'male' }

print(d1)
```

```python
# With mixed keys
d3 = {(1,2,3):1,'hello':'world'}

print(d3)
```

## **Dictionary Comprehension**
image.png

```python
# Print 1st 10 numbers and their squares
{i:i**2 for i in range(1,11)}
```

```python
# Using existing dict
distances = {'delhi':1000,'mumbai':2000,'bangalore':3000}
print(distances.items())

{key:value*0.62 for(key,value) in distances.items()}
```

```python
# Using zip
days = ["Sunday", "Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
temp_C = [30.5,32.6,31.8,33.4,29.8,30.2,29.9]

{i:j for(i,j) in zip(days,temp_C)}
```

```python
# Using if condition
products = {'phone':10,'laptop':0,'charger':32,'tablet':0}
{key:value for (key,value) in products.items() if value > 0}
```

```python
# Nested Comprehension --> Print tables of number from 2 to 4
{i:{j:i*j for j in range(1,11)}for i in range(2,5)}
```
