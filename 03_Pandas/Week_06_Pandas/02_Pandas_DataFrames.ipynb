# **DataFrames**

A DataFrame is a 2D labeled, mutable, heterogeneous tabular data structure provided by pandas, built on top of NumPy.

It is similar to:

- Excel sheet

- SQL table

- R data frame

Each column in a DataFrame is actually a Series object.

## **Filtering a DataFrame**

```python
# Preview the top two rows of the ipl matches dataset to check columns for filtering tasks
ipl.head(2)
```

```python
# Preview the first two rows of the movies DataFrame
movies.head(2)
```

```python
# Filter the ipl matches where the MatchNumber is 'Final', and extract only the Season and WinningTeam columns
mask = ipl['MatchNumber'] == 'Final'
new_df = ipl[mask]
new_df[['Season', 'WinningTeam']]
```

```python
# Filter the ipl DataFrame in a single line to display only Season and WinningTeam for 'Final' matches
ipl[ipl['MatchNumber'] == 'Final'][['Season', 'WinningTeam']]
```

```python
# Retrieve the number of rows (matches) where a Super Over was required (SuperOver == 'Y')
ipl[ipl['SuperOver'] == 'Y'].shape[0]
```

```python
# Count the number of matches won by Chennai Super Kings specifically when playing in Kolkata
ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]
```

```python
# Calculate the percentage of matches in which the team that won the toss also won the match
(ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0] / ipl.shape[0]) * 100
```

```python
# Count the number of movies that have an IMDb rating greater than 8 AND more than 1000 votes
movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 1000)].shape[0]
```

```python
# Filter and find action movies with a rating greater than 7.5 by checking list of split genre tokens
mask1 = movies['genres'].str.split('|').apply(lambda x:'Action' in x)
mask2 = movies['imdb_rating'] > 7.5
movies[mask1 & mask2]
```

```python
# Alternative filtering method using .str.contains() to identify 'Action' movies with ratings > 7.5
mask1 = movies['genres'].str.contains('Action')
mask2 = movies['imdb_rating'] > 7.5
movies[mask1 & mask2]
```

```python
# Write a function that can return the track record of 2 teams against each other
def team1_vs_team2(Team1, Team2):
    m1 = (ipl.Team1 == Team1) | (ipl.Team2 == Team1)
    m2 = (ipl.Team1 == Team2) | (ipl.Team2 == Team2)

    df1 = ipl[m1 & m2]

    return df1

team1_vs_team2('Chennai Super Kings', 'Mumbai Indians')
```
