## **2. Exceptions**

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime

- You have to takle is on the fly

**Examples**

- Memory overflow

- Divide by 0 -> Logical error

- Database error

### **Stack Trace**

Stack trace batata hai kis type ka error hai, kya galat hua, aur kis function / line se problem start hui.

A stack trace shows the error type, the error message, and the sequence of function calls that led to the error.

### **Why is it important to handle exceptions.**

Exception handling is important to improve user experience by avoiding crashes and to enhance security by hiding internal system details.





### **How to handle exceptions.**

Exceptions are handled using try, except, else, and finally blocks to prevent crashes and ensure graceful error handling.

#### **try except block**

**`try` – Risky Code**

Put the code that may cause an error inside try.

**`except` – Handle the Error**

Catch and handle the error without crashing the program

**`else` – Runs if No Exception Occurs**

**`finally` – Always Executes**

Runs whether error occurs or not

Used for cleanup (closing files, DB connections).

**Example** - External resources like files, databases, and networks can fail unexpectedly, so we use try–except–block to handle errors and release resources safely.

```python
# Let's create a file

with open('sample.txt','w') as f:
  f.write('Hello World')
```

```python
# try catch demo

try:
  with open('sample1.txt','r') as f:
    print(f.read())
except:
  print('Sorry file not found')
```

#### **Creating custom exceptions**

Creating custom exceptions means defining your own exception classes to handle application-specific or business-logic errors.

**Benefit of Creating Custom Exceptions**

- Full Control Over Errors - Custom exceptions give full control over how and when errors are raised and handled.

- Application Based Errors - Custom exceptions help represent business and application specific errors clearly.

```python
# Creating custom exceptions

class MyException(Exception):
  def __init__(self,message):
    print(message)

class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise MyException('Amount cannot be -ve')
    if self.balance < amount:
      raise MyException('Paise nai hai tere paas')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(5000)
except MyException as e:
  pass
else:
  print(obj.balance)
```

```python
# Simple Example - Google Wala Example

class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print('Logout')

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError('Bhai teri to lag gayi')
    if email == self.email and password == self.password:
      print('Welcome')
    else:
      print('Login error')


obj = Google('Nitish','nitish@gmail.com','1234','android')

try:
  obj.login('nitish@gmail.com','1234','android')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print('Database connection closed')
```
