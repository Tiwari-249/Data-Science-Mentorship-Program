```python
# Assign the integer 4 to the variable 'a'

a = 4
```

```python
# Get the memory address (ID) of the object 'a' refers to

id(a)
```

```python
# Get the memory address (ID) of the integer literal 4
# Note: In CPython, small integers (like 4) are often pre-allocated, so 'a' and '4' might refer to the same object ID

id(4)
```

```python
# Convert the integer memory address to its hexadecimal representation for easier viewing

hex(11126784)
```

```python
# Demonstrate aliasing: 'b' now refers to the same object as 'a'

a = 4
b = a
```

```python
# Print the memory addresses of 'a' and 'b'. They should be the same, confirming aliasing

print(id(a))
print(id(b))
```

```python
# Further demonstrate aliasing: 'c' now also refers to the same object as 'b' (and 'a')

c = b
```

```python
# Print the memory address of 'c'. It should be the same as 'a' and 'b'

print(id(c))
```

```python
# Delete the variable 'a'. This decreases the reference count of the object 4

del a
```

```python
# Trying to access 'a' after deletion will result in a NameError

a
```

```python
# 'b' still refers to the integer object 4, as its reference count is still greater than zero

print(b)
```

```python
# Delete the variable 'b'. 'c' still refers to the object 4

del b
print(c)
```

```python
# Demonstrate aliasing with a string object. 'a', 'b', and 'c' all reference the same string

a = 'DSMP 2022-23'
b = a
c = b
```

```python
# Use sys.getrefcount to see how many references point to the string object 'DSMP 2022-23'.
# This counts 'a', 'b', 'c', and the temporary reference created by sys.getrefcount itself

import sys
sys.getrefcount('DSMP 2022-23')
```

```python
# Demonstrate mutability of lists: A list 'L' is modified in-place, and its ID remains the same

L = [1,2,3]
print(id(L))

L.append(4)
print(L)
print(id(L))
```

```python
# Demonstrate immutability of tuples: A new tuple object is created when 'T' is 'modified'. The ID changes

T = (1,2,3)
print(id(T))

T = T + (4,)

print(T)
print(id(T))
```

```python
# Further demonstration of list aliasing: modifying 'b' (which is an alias of 'a') also changes 'a'

a = [1,2,3]
b = a

b.append(4)
print(b)
```

```python
# 'a' also reflects the change made via 'b' because they both refer to the same list object.
print(a)
```

```python
# Demonstrate how passing a slice (which creates a new list) to a function prevents side effects on the original list

def func(data):
  data.append(4)

a = [1,2,3]
# func(a[:]) passes a shallow copy of 'a', so 'a' itself is not modified
func(a[:])
print(a)
```

```python
# Demonstrate cloning (shallow copy for a simple list) using slicing

a = [1,2,3]

# cloning: 'b' is a new list object with the same elements as 'a'
b = a[:]
```

```python
# Get the ID of list 'a'

id(a)
```

```python
# Get the ID of list 'b'. It should be different from 'a', confirming it's a new object

id(b)
```

```python
# Modify list 'b'. Since 'b' is a separate object, this modification will not affect 'a'

b.append(4)
b
```

```python
# 'a' remains unchanged, demonstrating that slicing creates a separate list object (a shallow copy)

a
```

```python
# Demonstrate shallow copying with dictionaries using the `.copy()` method. This also creates a new dictionary object

a = {'name':'nitish','age':33}

# aliasing->cloning (this comment seems to imply both, but `copy()` is cloning/shallow copy)
b = a.copy()

b['gender'] = 'male'

print(b)
print(a)
```

```python
# Demonstrate shallow copying with lists using `.copy()`. For a list of immutable elements, it behaves like a deep copy

a = [1,2,3]

# shallow copy
b = a.copy()

b.append(4)

print(a)
print(b)
```

```python
# Introduce a nested list to demonstrate the limitation of shallow copies

a = [1,2,3,[4,5]]
a
```

```python
# Check the IDs of the nested lists in 'a' and 'b'. For a shallow copy, these IDs will be the same
# Note: This cell might need 'b' to be defined as a copy of 'a' first, if not already

print(id(a[-1]))
print(id(b[-1]))
```

```python
# Create a shallow copy of 'a' using .copy()

b = a.copy()
b
```

```python
# Print the IDs of the outer lists 'a' and 'b'. They should be different

print(id(a))
print(id(b))
```

```python
# Modify an element within the NESTED list of 'b'
# Because it's a shallow copy, the nested list object is shared between 'a' and 'b'

b[-1][0] = 400
print(b)
```

```python
# Observe that 'a' also reflects the change in its nested list, highlighting the 'danger' of shallow copies with mutable nested objects

a
```

```python
# Import the `copy` module, which is used for both shallow and deep copies

import copy
a = [1,2,3,[4,5]]
a
```

```python
# Use slicing to create a shallow copy 'b' from 'a'

b = a[:]
b
```

```python
# Modify an element within the nested list of 'b'

b[-1][0] = 400
b
```

```python
# 'a' is also modified, confirming that slicing for nested mutable structures results in a shallow copy

a
```

```python
# Check the IDs of the nested lists in 'a' and 'b'. They are the same because it's a shallow copy

print(id(a[-1]))
print(id(b[-1]))
```
