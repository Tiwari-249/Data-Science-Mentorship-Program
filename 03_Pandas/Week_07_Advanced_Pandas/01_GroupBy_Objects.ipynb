# **Groupby Objects**

Groupby Objects - Study of the groups.

In Pandas, a GroupBy object is created when you use the .groupby() method on a DataFrame or Series.

It follows the `Split → Apply → Combine` strategy.

```python
# Import necessary libraries for data analysis and mathematical operations
import numpy as np
import pandas as pd
```

```python
# Load the IMDB Top 1000 dataset from Google Drive into a Pandas DataFrame
movies = pd.read_csv('/content/drive/MyDrive/DataSets/Session 19/imdb-top-1000.csv')
```

```python
# Check the dimensions (rows, columns) of the loaded movies DataFrame
movies.shape
```

```python
# Display the first 5 rows of the movies dataset to inspect the data
movies.head(5)
```

```python
# Group the movies dataset by the 'Genre' column
genres = movies.groupby('Genre')
```

```python
# Apply the built-in sum aggregation function on the GroupBy object
# Note: This will concatenate strings for text columns and sum numerical values
genres.sum()
```

```python
# Find the minimum value for each column across all groups in the genres GroupBy object
genres.min()
```

```python
# Calculate the mean value of only numerical columns for each movie genre
genres.mean(numeric_only=True)
```

```python
# Calculate the standard deviation of only numerical columns for each movie genre
genres.std(numeric_only=True)
```

```python
# Find the top 3 genres by total earning (Gross value) by aggregating, sorting, and fetching head
movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)
```

```python
# A faster way to find the top 3 genres by selecting only the 'Gross' column before summing
movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3)
```

```python
# Find the genre with the highest average IMDB rating
movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
# Find the top 3 directors based on the total sum of votes their movies received
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(3)
```

```python
# Find the highest IMDB rating scored by a movie in each genre
movies.groupby('Genre')['IMDB_Rating'].max()
```

```python
# Count the number of movies starred in by each lead actor (Star1)
movies.groupby('Star1')['Series_Title'].count()
```

### **Agg Method**

```python
# Aggregate columns by passing a dictionary to specify different functions for different columns
genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum',
        'Metascore':'min'
    }
)
```

```python
# Select numeric columns and apply a list of aggregation functions across all of them
numeric_cols = movies.select_dtypes(include=np.number).columns
genres[numeric_cols].agg(['min','max','mean','sum'])
```

```python
# Combine dictionary and list syntax to apply multiple distinct aggregations per column
genres.agg(
    {
        'Runtime':['mean','min'],
        'IMDB_Rating':'mean',
        'No_of_Votes':['sum','max'],
        'Gross':'sum',
        'Metascore':'min'
    }
)
```

```python
# Iterate through the GroupBy object to loop and print the sub-DataFrame for each genre group
for groups,data in genres:
    print(data)
```

```python
# Iterate through each group to find the row with the maximum IMDB rating and concatenate them
results = []
for group,data in genres:
    results.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])

df = pd.concat(results)
df
```

### **split (apply) combine**

`apply` -> builtin function

```python
# Apply the built-in 'min' function over each group using the apply method
genres.apply(min, include_groups=False)
```

```python
# Define a custom helper function to find the total count of movie titles starting with the letter 'A'
def foo(group):
    return group['Series_Title'].str.startswith('A').sum()
```

```python
# Apply the custom 'foo' function to calculate the frequency of titles starting with 'A' in each genre
genres.apply(foo)
```

```python
# Define a helper function to assign a rank to each movie within its genre based on its IMDB Rating
def rank_movie(group):
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False)
    return group
```

```python
# Apply the movie ranking function across all genre groups to get a rank-added DataFrame
genres.apply(rank_movie)
```

```python
# Define a helper function to apply min-max normalization on the IMDB rating within each group
def normal(group):
    group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) / (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group
```

```python
# Apply the custom normalization helper function over all genre groups
genres.apply(normal)
```

```python
# Group movies by multiple columns: 'Director' and 'Star1' and perform basic inspections
duo = movies.groupby(['Director','Star1'])

# Size of each director-actor combo
duo.size()

# Retrieve records for the specific combination: Aamir Khan and Amole Gupte
duo.get_group(('Aamir Khan','Amole Gupte'))
```

```python
# Find the director-actor pair that generated the highest cumulative Gross earnings
duo['Gross'].sum().sort_values(ascending=False).head(1)
```

```python
# Find the actor-genre pair with the highest average Metascore rating
movies.groupby(['Star1','Genre'])['Metascore'].mean().reset_index().sort_values('Metascore',ascending=False).head(1)
```

```python
# Apply min, max, and mean aggregation functions on numerical columns for director-actor groups
numeric_cols = movies.select_dtypes(include=np.number).columns
duo[numeric_cols].agg(['min','max','mean'])
```

## **Excercise**

```python
# Load the IPL deliveries dataset from Google Drive into a DataFrame for the exercise section
ipl = pd.read_csv('/content/drive/MyDrive/DataSets/Session 19/deliveries.csv')
```

```python
# Display the first 5 records of the loaded IPL deliveries DataFrame to check columns
ipl.head()
```

```python
# Find total rows and columns present in the IPL deliveries dataset
ipl.shape
```

```python
# Identify the top 10 batsmen based on total runs scored throughout the IPL history
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# Filter deliveries with 6 runs, group by batsman, count them, and find who hit the most sixes
six = ipl[ipl['batsman_runs'] == 6]
six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
# Find the batsman with the most boundaries (fours/sixes) in the death overs (over > 15)
temp_df = ipl[ipl['over'] > 15]
temp_df = temp_df[(temp_df['batsman_runs'] == 4) | (temp_df['batsman_runs'] == 6)]
temp_df.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
# Extract and group total runs scored by Virat Kohli against all opposing teams
temp_df = ipl[ipl['batsman'] == 'V Kohli']
temp_df.groupby('bowling_team')['batsman_runs'].sum().reset_index()
```

```python
# Find Virat Kohli's single highest score in any match by grouping his runs by match_id
temp_df = ipl[ipl['batsman'] == 'V Kohli']
temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
# Create a generic function to calculate and return the highest score of any given batsman
def highest(batsman):
    temp_df = ipl[ipl['batsman'] == batsman]
    return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
# Call the generic function to find the highest score of V Kohli
highest('V Kohli')
```

```python
# Call the generic function to find the highest score of MS Dhoni
highest('MS Dhoni')
```

```python
# Call the generic function to find the highest score of CH Gayle
highest('CH Gayle')
```
