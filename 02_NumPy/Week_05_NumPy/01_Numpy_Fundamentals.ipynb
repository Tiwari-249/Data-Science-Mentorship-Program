# **What is numpy?**

NumPy is the fundamental package for scientific computing in Python. It is a Python library that provides a multidimensional array object, various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays, including mathematical, logical, shape manipulation, sorting, selecting, I/O, discrete Fourier transforms, basic linear algebra, basic statistical operations, random simulation and much more.


At the core of the NumPy package, is the ndarray object. This encapsulates n-dimensional arrays of homogeneous data types.

## **Numpy Arrays Vs Python Sequences**

- NumPy arrays have a fixed size at creation, unlike Python lists (which can grow dynamically). Changing the size of an ndarray will create a new array and delete the original.

- The elements in a NumPy array are all required to be of the same data type, and thus will be the same size in memory.

- NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python’s built-in sequences.

- A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typically support Python-sequence input, they convert such input to NumPy arrays prior to processing, and they often output NumPy arrays.

## **Creating Numpy Arrays**

```python
# Create a 1D NumPy array (vector) from a list
a = np.array([1,2,3])
print(a)
```

## **Stacking**

Stacking means joining multiple arrays along a specified axis.

NumPy stacking functions like `hstack` and `vstack` are used to combine arrays horizontally or vertically, provided their dimensions are compatible.

**Uses** - Stacking is used to combine multiple arrays for data preparation, feature engineering, and matrix construction

```python
# Create and print array a4 for stacking examples
a4 = np.arange(12).reshape(3,4)
print(a4)
```

```python
# Create and print array a5 for stacking examples
a5 = np.arange(12,24).reshape(3,4)
print(a5)
```

### **Horizontal Stacking** — `hstack()`

Joins arrays side-by-side (columns).

```python
# Stack arrays horizontally (side-by-side along columns)
np.hstack((a4,a5))
```

### **Vertical Stacking** — `vstack()`

Joins arrays one below another (rows).

```python
# Stack arrays vertically (one on top of the other along rows)
np.vstack((a4,a5))
```

Spliting means dividing an array into multiple sub-array.

NumPy provides `hsplit` and `vsplit` to split arrays column-wise or row-wise, while array_split allows unequal divisions.

**Uses** - Splitting is used to divide arrays for batch processing, train-test splits, and parallel computation.

```python
# Print array a4 to show its structure before splitting
print(a4)
```

```python
# Print array a5 to show its structure before splitting
print(a5)
```

### **Horizontal Split** — `hsplit()`

Splits array column-wise.

Columns must be divisible.

```python
# Split array a4 horizontally into 2 equal column-wise sub-arrays
np.hsplit(a4,2)
```

```python
# Split array a4 horizontally into 4 equal column-wise sub-arrays
np.hsplit(a4,4)
```

```python
# Attempt to split horizontally into 5 parts (Will fail because columns must be evenly divisible)
np.hsplit(a4,5)
```

### **Vertical Split** — `vsplit()`

Splits array row-wise.

Rows must be divisible.

```python
# Split array a5 vertically into 3 equal row-wise sub-arrays
np.vsplit(a5,3)
```

```python
# Attempt to split vertically into 2 parts (Will fail because rows must be evenly divisible)
np.vsplit(a5,2)
```
